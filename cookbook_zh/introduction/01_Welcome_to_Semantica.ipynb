Semantica 是一个**语义智能（Semantic Intelligence）与知识工程（Knowledge Engineering）框架**。它帮助你：

- 从非结构化（Unstructured）和半结构化数据构建**知识图谱（Knowledge Graph）**
- 在多样化数据源之上建立统一的**语义层（Semantic Layer）**
- 用结构化知识驱动 **GraphRAG**、AI 智能体（Agent）和多智能体系统
- 在应用中引入**时态与质量感知推理（Reasoning）**

### 核心能力

- **通用摄取（Ingestion）**：文件、网页、Feed、数据库、仓库、流
- **丰富解析（Parsing）**：PDF、Office 文档、HTML、JSON、CSV、图像、代码
- **规范化（Normalization）**：清洗、语言检测、实体归一化、日期/数字标准化
- **语义抽取（Semantic Extraction）**：命名实体识别（NER）、关系、事件、语义网络
- **知识图谱构建**：由实体和关系构建属性图（Property Graph）
- **嵌入（Embedding）与向量检索**：文本与图嵌入、混合检索（Hybrid Retrieval）
- **推理与本体（Ontology）**：基于规则的推断、本体生成与校验
- **可视化与分析**：图可视化和质量指标

## Semantica 适合谁？

- 构建 GraphRAG 系统、智能体和需要长期记忆工具的 **AI/ML 工程师**
- 在大型异构数据集上编排语义增强流水线的**数据工程师**
- 设计并维护形式化知识结构的**知识工程师与本体专家（Ontologist）**
- 从文档和数据 Feed 构建领域知识图谱的**研究人员与分析师**
- 将语义智能嵌入应用与服务的**产品与平台团队**

## 架构总览

Semantica 由三个概念层和多个具体模块组成。

### 分层

- **输入层（Input Layer）**
  - 对接文件、网页、API、数据库、邮件、Feed、仓库和流
  - 将这些不同来源规范化为统一的内部表示

- **语义层（Semantic Layer）**
  - 执行解析、清洗、语义抽取、图构建、嵌入和推理
  - 这里是**非结构化数据变成结构化知识**的地方

- **输出层（Output Layer）**
  - 对外提供知识图谱、嵌入、本体和分析结果
  - 与向量库、图数据库及下游应用集成

## 🧩 Semantica 模块参考

Semantica 采用模块化设计。以下按功能分组，给出所有可用模块的完整指南。

### 📥 摄取与解析
处理原始数据输入和结构的模块。

| 模块 | 说明 | 关键能力 |
| :--- | :--- | :--- |
| **`ingest`** | **数据摄取（Data Ingestion）**<br>对接数据源。 | • 文件、Web、Feed、流式摄取<br>• 数据库、邮件、仓库、MCP 支持 |
| **`parse`** | **文档解析（Document Parsing）**<br>把原始内容解析为结构。 | • PDF、HTML、JSON、CSV、Excel<br>• 图像与代码解析 |

### ⚙️ 数据处理
负责清洗、规范化和拆分数据的模块。

| 模块 | 说明 | 关键能力 |
| :--- | :--- | :--- |
| **`normalize`** | **数据规范化（Data Normalization）**<br>清洗并标准化文本。 | • 文本清洗与语言检测<br>• 实体、日期、数字规范化 |
| **`split`** | **分块（Chunking）**<br>为 RAG 拆分文档。 | • 递归字符拆分<br>• 语义与基于 Token 的拆分 |

### 🧠 抽取与增强
从原始数据中抽取含义、结构和向量的模块。

| 模块 | 说明 | 关键能力 |
| :--- | :--- | :--- |
| **`semantic_extract`** | **信息抽取（Information Extraction）**<br>抽取实体和关系。 | • NER 与关系抽取<br>• 事件与语义网络检测 |
| **`context`** | **智能体记忆（Agent Memory）**<br>管理 AI 智能体的状态。 | • 长期记忆与历史<br>• 上下文图（Context Graph）与 RAG 集成 |

### 🕸️ 知识图谱核心
构建、精炼和消解知识图谱的模块。

| 模块 | 说明 | 关键能力 |
| :--- | :--- | :--- |
| **`kg`** | **图构建（Graph Construction）**<br>构建和分析图。 | • 图构建与分析<br>• 校验与实体消解（Entity Resolution） |
| **`conflicts`** | **冲突消解（Conflict Resolution）**<br>处理数据矛盾。 | • 来源可靠性评分<br>• 真相发现（Truth Discovery）算法 |
| **`deduplication`** | **实体消解**<br>合并重复实体。 | • 基于相似度的分块<br>• 聚类与规范化 |

### 💾 存储与检索
持久化与查询数据的模块。

| 模块 | 说明 | 关键能力 |
| :--- | :--- | :--- |
| **`embeddings`** | **向量嵌入（Vector Embedding）**<br>生成语义向量。 | • 文本与图嵌入<br>• 多提供商支持（OpenAI 等） |
| **`vector_store`** | **向量数据库（Vector Database）**<br>存储和检索向量。 | • 相似度检索与过滤<br>• 混合检索（向量 + 关键词） |
| **`graph_store`** | **属性图存储（Property Graph Store）**<br>持久化图数据。 | • Neo4j、FalkorDB 适配器<br>• Cypher 查询支持 |
| **`triplet_store`** | **RDF 存储**<br>持久化语义三元组（Triplet）。 | • SPARQL 端点<br>• BlazeGraph、Jena、Virtuoso 适配器 |

### 🔎 推理与分析
推导新知识和评估质量的模块。

| 模块 | 说明 | 关键能力 |
| :--- | :--- | :--- |
| **`reasoning`** | **推理门面（Reasoner Facade）**<br>统一的推理接口。 | • Datalog/基于规则的推理<br>• 前向/后向链（Forward/Backward Chaining） |
| **`ontology`** | **本体管理（Ontology Management）**<br>管理模式和定义。 | • 从数据生成本体<br>• 校验与演化 |
| **`visualization`** | **可视化分析（Visual Analytics）**<br>可视化图与指标。 | • 2D/3D 图可视化<br>• 交互式图表与仪表盘 |
| **`evals`** | **评估（Evaluation）**<br>对流水线质量做基准测试。 | • RAG 与图质量指标<br>• 与真值（Ground Truth）对比 |

### 🛠️ 编排与工具
管理框架与工作流的模块。

| 模块 | 说明 | 关键能力 |
| :--- | :--- | :--- |
| **`core`** | **框架核心（Framework Core）**<br>主入口与配置。 | • 生命周期管理<br>• 插件系统与配置 |
| **`pipeline`** | **工作流编排（Workflow Orchestration）**<br>管理复杂流程。 | • DAG 执行与重试<br>• 错误处理与可观测性 |
| **`seed`** | **数据播种（Data Seeding）**<br>初始化知识库。 | • 分类法（Taxonomy）与本体播种<br>• 参考数据加载 |
| **`export`** | **数据导出（Data Export）**<br>把数据导出为文件。 | • JSON、CSV、RDF、GEXF 导出<br>• 报告生成 |
| **`utils`** | **工具集（Utilities）**<br>通用辅助函数。 | • 日志、异步、哈希<br>• 文本处理辅助 |

## 核心概念（高层）

- **知识图谱（Knowledge Graph）**
  - 节点表示实体（Entity），如人物、组织、地点、事件或概念
  - 边表示关系（Relationship），如 `works_for`、`located_in`、`founded_by`
  - 属性记录属性和元数据（Metadata），如时间戳、来源和置信度（Confidence）

- **实体与关系**
  - 实体通过 NER 从文本和数据中抽取
  - 关系连接实体，通过基于模式、基于模型或基于 LLM 的方法抽取

- **嵌入（Embedding）**
  - 编码文本或图结构语义含义的数值向量
  - 用于语义搜索、聚类和基于相似度的检索

- **GraphRAG**
  - 结合向量检索与图遍历（Graph Traversal）
  - 同时利用嵌入和图结构来检索丰富的、具备上下文感知的信息

- **本体（Ontology）**
  - 领域内类、关系和约束的形式化模型
  - 用于统一语义、启用推理并集成异构数据

- **质量与治理（Quality and Governance）**
  - 质量指标：完整性、一致性、准确性、覆盖度
  - 知识图谱层面的冲突检测与消解

## 准备：创建示例数据

首先，创建一个小的示例文档来练手。

````python
from pathlib import Path

docs_dir = Path("welcome_docs")
docs_dir.mkdir(exist_ok=True)
text_path = docs_dir / "apple.txt"
text_content = (
    "Apple Inc. was founded by Steve Jobs, Steve Wozniak and Ronald Wayne in"
    " Cupertino, California."
)
text_path.write_text(text_content, encoding="utf-8")
print(f"Created sample document at {text_path}")
````

## 最小端到端流水线（Pipeline）

下一个示例展示如何按顺序显式使用多个模块。它呼应了前面讨论的架构：

1. 摄取一个文档目录
2. 把它们解析为结构化文档
3. 规范化文本
4. 抽取实体和关系
5. 构建并分析知识图谱
6. 创建嵌入并存入向量库
7. 执行一次混合语义检索查询

````python
from semantica.ingest import FileIngestor
from semantica.parse import DocumentParser
from semantica.normalize import TextNormalizer
from semantica.semantic_extract import NERExtractor, RelationExtractor
from semantica.kg import GraphBuilder, GraphAnalyzer
from semantica.embeddings import EmbeddingGenerator
from semantica.vector_store import VectorStore, HybridSearch

ingestor = FileIngestor()
documents = ingestor.ingest(str(docs_dir))

parser = DocumentParser()
parsed_docs = parser.parse(documents)

normalizer = TextNormalizer()
normalized_docs = normalizer.normalize(parsed_docs)

ner = NERExtractor()
entities = ner.extract(normalized_docs)
rel_extractor = RelationExtractor()
relationships = rel_extractor.extract(normalized_docs, entities)

builder = GraphBuilder()
kg = builder.build(entities, relationships)
analyzer = GraphAnalyzer()
metrics = analyzer.analyze(kg)

emb_generator = EmbeddingGenerator()
embeddings = emb_generator.generate_embeddings(documents, data_type="text")

vec_store = VectorStore(dimension=384)  # 与 EmbeddingGenerator 默认嵌入维度(384)对齐
vec_store.store(embeddings, documents, metadata={})
hybrid = HybridSearch(vec_store)
search_results = hybrid.search("Apple founders", top_k=3)
len(search_results)
````

## 方法对比:模式抽取 vs LLM 抽取

上面流水线用的 `RelationExtractor()` 默认是**模式(Pattern)方法**——预置正则句式。
对照句子"…founded by Steve Jobs, Steve Wozniak and Ronald Wayne…",
正则只抓到 `founded by` 后的第一个实体名,**逗号并列的其余创始人被丢弃**。

下面用**LLM 方法**(智谱 GLM-5.3-flash,OpenAI 兼容协议)重新抽取同一段文本,
对比两者的关系抽取质量。前置条件:`config.yaml` 里填入你的智谱 API Key
(或设置环境变量 `ZHIPU_API_KEY`)。

````python
import os
import yaml
from semantica.semantic_extract import RelationExtractor

# ── API Key:环境变量与 config.yaml 都试,取第一个合法值 ──
cfg = yaml.safe_load(Path("config.yaml").read_text(encoding="utf-8"))
candidates = {
    "环境变量 ZHIPU_API_KEY": os.environ.get("ZHIPU_API_KEY"),
    "config.yaml": cfg["llm_provider"]["api_key"],
}
print(candidates) 
api_key = next((k for k in candidates.values() if k and not k.startswith("<")), "")
assert api_key, (
    "未找到合法的智谱 API Key。各来源状态: "
    + "; ".join(
        f"{name}={'占位符' if v and v.startswith('<') else ('空' if not v else '有值')}"
        for name, v in candidates.items()
    )
    + "。注意:若内核启动晚于变量设置,需重启 Jupyter 内核。"
)

# relationships 是按文档分组的嵌套列表(extract 对 list 输入逐文档返回),先展平
def _flatten(rels):
    out = []
    for item in rels:
        out.extend(item) if isinstance(item, list) else out.append(item)
    return out

pattern_rels = _flatten(relationships)
print(f"pattern 抽出 {len(pattern_rels)} 条:")
for r in pattern_rels:
    print(f"  {r.subject.text} -[{r.predicate}]-> {r.object.text}")

# ── 方法二:LLM(智谱 GLM-5.3-flash,OpenAI 兼容协议)──
llm_extractor = RelationExtractor(
    method="llm",
    provider="openai",
    llm_model="glm-5.3-flash",
    base_url="https://open.bigmodel.cn/api/coding/paas/v4",
    api_key=api_key,
)
llm_rels = _flatten(llm_extractor.extract(normalized_docs, entities))

print(f"\nLLM 抽出 {len(llm_rels)} 条:")
for r in llm_rels:
    print(f"  {r.subject.text} -[{r.predicate}]-> {r.object.text}  (conf={r.confidence})")
````

## 可视化:两次抽取的左右对比

把两种方法对同一句话的关系抽取结果并排画出——左图 **Pattern 方法(正则)**,
右图 **LLM 方法(GLM-5.3-flash)**。孤立点 = 实体被 NER 认出但没有任何关系边相连。

> 上游原演示(KGVisualizer 模块画单图)见英文原版 notebook;此处替换为对比视图。
> 注意:若智谱调用因余额失败,右图会退化成兜底结果(pattern + 相邻连接),
> 此时两图差异不大——这是方法的局限,不是代码问题。

````python
import math
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def _dedupe(rels):
    """按 (主体, 谓语, 客体) 三元组去重。"""
    seen, out = set(), []
    for r in rels:
        key = (r.subject.text, r.predicate, r.object.text)
        if key not in seen:
            seen.add(key)
            out.append(r)
    return out

def _collect_nodes(rels, base_pos):
    """收集关系中出现、但基准坐标里没有的节点名。"""
    names = []
    for r in rels:
        for n in (r.subject.text, r.object.text):
            if n not in base_pos and n not in names:
                names.append(n)
    return names

BASE_POS = {"Apple Inc.": (0.0, 0.0)}          # 主体固定在下方中央

def draw(fig, rels, ents, col, title):
    rels = _dedupe(rels)
    # 节点全集 = NER 实体 + 关系端点 —— 孤立实体(无关系边)也要画出来,
    # 它正是 pattern 方法丢关系的直接证据
    names = [e.text for e in ents]
    for r in rels:
        for n in (r.subject.text, r.object.text):
            if n not in names:
                names.append(n)
    pos = dict(BASE_POS)
    extra = [n for n in names if n not in pos]
    for i, n in enumerate(extra):              # 新节点按弧线自动排布
        ang = math.pi * (0.15 + 0.7 * i / max(len(extra) - 1, 1))
        pos[n] = (-2.4 * math.cos(ang), 1.9 * math.sin(ang) + 0.7)
    for r in rels:                             # 边:线段 + 中点谓语标签
        s, o = r.subject.text, r.object.text
        if s in pos and o in pos:
            (x1, y1), (x2, y2) = pos[s], pos[o]
            fig.add_trace(go.Scatter(
                x=[x1, x2], y=[y1, y2], mode="lines",
                line=dict(width=1.6, color="#8fa8c6"),
                hovertext=f"{s} -[{r.predicate}]-> {o}",
                hoverinfo="text", showlegend=False,
            ), row=1, col=col)
            fig.add_trace(go.Scatter(
                x=[(x1 + x2) / 2], y=[(y1 + y2) / 2], mode="text",
                text=[r.predicate], textfont=dict(size=10, color="#5b7a9d"),
                hoverinfo="skip", showlegend=False,
            ), row=1, col=col)
    fig.add_trace(go.Scatter(                  # 节点
        x=[p[0] for p in pos.values()], y=[p[1] for p in pos.values()],
        mode="markers+text", text=list(pos.keys()), textposition="top center",
        textfont=dict(size=11),
        marker=dict(size=15, color=["#26a69a" if n == "Apple Inc." else "#ef5350" for n in pos]),
        showlegend=False,
    ), row=1, col=col)
    fig.update_xaxes(visible=False, row=1, col=col)
    fig.update_yaxes(visible=False, row=1, col=col)

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Pattern 方法（正则）", "LLM 方法（GLM-5.3-flash）"),
)
draw(fig, _flatten(relationships), entities, 1, "pattern")
draw(fig, _flatten(llm_rels), entities, 2, "llm")
fig.update_layout(
    height=460, showlegend=False,
    title_text="同一句话的两种关系抽取结果对比",
    plot_bgcolor="rgba(0,0,0,0)", paper_bgcolor="rgba(0,0,0,0)",
)
fig.show()
````

## 进阶：推理与推断

`reasoning` 模块让你能用逻辑规则从已有知识推导出新事实。

````python
from semantica.reasoning import Reasoner

# Simple rule: If X founded Y, then X works_for Y
rule = """
IF (?x founded ?y) THEN (?x works_for ?y)
"""

reasoner = Reasoner()
reasoner.add_rule(rule)
inferred_facts = reasoner.infer_facts(kg)

print(f"Inferred {len(inferred_facts)} new facts")
````

## 使用核心 `Semantica` 类

````python
from semantica.core import Semantica, ConfigManager

config_manager = ConfigManager()
config = config_manager.load_from_file("config.yaml")

framework = Semantica(config=config)
framework.initialize()

kb_result = framework.build_knowledge_base(
    sources=[str(docs_dir)],
    embeddings=True,
    graph=True,
)

framework.shutdown()
sorted(kb_result.keys())
````
